```python
#to check whether or not the traingle is a right angeled traingle using function
def right_traingle(a,b,c):
    if a*a + b*b == c*c:
        print("Right angeled traingle")
    else:
         print("not a right angeled traingle") 
right_traingle(3,4,5)
```
