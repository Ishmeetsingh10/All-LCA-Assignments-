```python
#reverse the number
num = int(input("enter a number:"))
reverse = 0
while num > 0:
    digit = num %10
    reverse = digit + reverse*10
    num = num//10

print("reverse=",reverse)
```

```python
#find wheather number is leap year
year = int(input("enter a year:"))

if year % 400 ==0:
    print("the year is a leap year")
elif year % 100 ==0:
    print("the year is not a leap year")
elif year % 4 ==0:
    print ("the year is a leap year")
else:
    print("the year is not a leap year")
```
